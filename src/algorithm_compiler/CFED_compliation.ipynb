```python
import os
import glob

def get_algorithm_details(path):
    """
    Scans the current directory and subdirectories for algorithm files and extracts their details.
    Returns a list of dictionaries containing algorithm path, name and function information.
    """
    algorithms = []
    
    # Walk through all directories
    for root, dirs, files in os.walk(path):
        # Look for .cpp files but skip temp_algorithm.cpp files
        for file in files:
            if file.endswith('.cpp') and file != 'temp_algorithm.cpp':
                filepath = os.path.join(root, file)
                
                # Read the file content
                with open(filepath, 'r') as f:
                    content = f.read()
                    
                algo_info = {
                    'path': os.path.abspath(filepath),
                    'filename': file,
                    'functions': []
                }
                
                # Basic function detection - can be enhanced for more robust parsing
                lines = content.split('\n')
                for line in lines:
                    # Look for function definitions
                    if ('void' in line or 'int' in line or 'char' in line) and '(' in line and ')' in line and '{' in line:
                        # Extract function name
                        func_name = line.split('(')[0].split()[-1]
                        algo_info['functions'].append(func_name)
                
                if algo_info['functions']:  # Only add if functions were found
                    algorithms.append(algo_info)
    
    return algorithms

# Example usage
algorithms = get_algorithm_details("/workspace/data/C-Plus-Plus")

print(algorithms)

# Print details of first algorithm if any exist
if algorithms:
    print("\nFirst Algorithm Details:")
    print(f"Filename: {algorithms[0]['filename']}")
    print(f"Full Path: {algorithms[0]['path']}")
    print("Functions:")
    for func in algorithms[0]['functions']:
        print(f"  - {func}")
else:
    print("No algorithms found")

# for algo in algorithms:
#     print(f"\nAlgorithm File: {algo['filename']}")
#     print(f"Full Path: {algo['path']}")
#     print("Functions:")
#     for func in algo['functions']:
#         print(f"  - {func}")
```

```python
from build_archiver import BuildArchiver
#from sqlite_archiver import SQLiteBuildArchiver
import sys
import io
from contextlib import redirect_stdout, redirect_stderr

technique = "RACFED"
technique_type = "SigMon"
selective_level = 0

archiver = BuildArchiver()
# archiver = SQLiteBuildArchiver()

# clean up the archive
# archiver.clean_archive(force=True)

# Import tqdm for progress bar visualization
from tqdm import tqdm

# Create a progress bar for algorithm compilation
print("\nCompiling algorithms with progress tracking:")
algorithms_to_compile = tqdm(algorithms[69:], desc="Compiling Algorithms", unit="algorithm")

# Redirect stdout/stderr to capture debug output
stdout_capture = io.StringIO()
stderr_capture = io.StringIO()

for algorithm in algorithms_to_compile:
    # Capture all output during compilation
    with redirect_stdout(stdout_capture), redirect_stderr(stderr_capture):
        compile_algorithm(algorithm, technique, technique_type, selective_level)
        
        
        
        # Archive the build
        archiver.archive_plugin_output(
            build_dir=os.path.join(os.getcwd(), 'BUILD'),
            algorithm_info=algorithm,
            compilation_params={
                'technique': technique,
                'technique_type': technique_type,
                'selective_level': selective_level
            }
        )
        print("Build archived successfully")
    
    # Only print captured output if there were errors
    if "failed" in stdout_capture.getvalue().lower() or stderr_capture.getvalue():
        print("\nCompilation output:")
        print(stdout_capture.getvalue())
        print(stderr_capture.getvalue())
        
    # Clear the buffers
    stdout_capture.truncate(0)
    stdout_capture.seek(0)
    stderr_capture.truncate(0) 
    stderr_capture.seek(0)
```
