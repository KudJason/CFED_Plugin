# Inspect Fine-tuning Data (Parquet)

This notebook loads and inspects the fine-tuning data exported in Parquet format.

```python
# Install necessary libraries if not already installed
%pip install pandas pyarrow
```

```python
import pandas as pd
import json # For potentially parsing JSON strings within columns

# Define the path to the Parquet file
# Adjust the path if your notebook or data file is in a different location
parquet_file_path = '/workspace/data/finetune_data.parquet' # Relative path from src/analyzer
parquet_file_path
```

```python
# Load the Parquet file into a pandas DataFrame
try:
    df = pd.read_parquet(parquet_file_path)
    print(f"Successfully loaded {len(df)} records from {parquet_file_path}")
except FileNotFoundError:
    print(f"Error: File not found at {parquet_file_path}. Please check the path.")
except Exception as e:
    print(f"An error occurred while loading the Parquet file: {e}")
```

```python
# Display the first few rows of the DataFrame
if 'df' in locals():
    display(df.head())
```

```python
# Display basic information about the DataFrame (columns, data types, non-null counts)
if 'df' in locals():
    df.info()
```

```python
# Display summary statistics for all columns (including categorical)
if 'df' in locals():
    display(df.describe(include='all'))
```

```python
# Check the distribution of different instruction types
if 'df' in locals():
    print("\nInstruction Distribution:")
    display(df['instruction'].value_counts())
```
